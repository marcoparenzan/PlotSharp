# MàthemaNet.LinearAlgebra — Usage: LU
Kernel C# .NET 11. Compilare in Release prima dell'esecuzione. Calcoli nello stesso tipo T, senza conversioni intermedie in double.


```
#load "../shared/MathemaSharp.csx"
using MathemaSharp.LinearAlgebra;
using System.Numerics;
void Example<T>() where T : IFloatingPointIeee754<T> {
    T[,] a = {{T.CreateChecked(2), T.Zero}, {T.Zero, T.CreateChecked(4)}};
    var lu = LuFactorization<T>.Create(a);
    var x = lu.Solve([T.CreateChecked(2), T.CreateChecked(8)]);
    Console.WriteLine($"{typeof(T).Name}: [{x[0]}, {x[1]}]");
}
Example<double>();
Example<Decimal32>();
Example<Decimal64>();
Example<Decimal128>();
```
